#### Question 1
#### Employee Data Bank.



Make a class called Employee. The __init__() method for Employee should store its different attributes: name, number, address, salary, post, year of joining, education, and his year of experience.

Make a method called describe_employee() that prints the detail Information about employee, a method update_ year_of_experience(), and a method called increase_salary() that do increamnet in it's salary.

Make an instance called employee101 from your class. Print the four attri-
butes individually, and then call all it's methods for describing employee, ssalary increamnet and increase year of experience.

Create 10 different instances from the class, and call describe_employee() for each
instance. Then do some updations with it's methods.



```python
class Employee:
    def __init__(self, name, number, address, salary, post, year_of_joining, education, year_of_experience):
        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.year_of_joining = year_of_joining
        self.education = education
        self.year_of_experience = year_of_experience

    def describe_employee(self):
        print(f"Name: {self.name}")
        print(f"Number: {self.number}")
        print(f"Address: {self.address}")
        print(f"Salary: {self.salary}")
        print(f"Post: {self.post}")
        print(f"Year of Joining: {self.year_of_joining}")
        print(f"Education: {self.education}")
        print(f"Year of Experience: {self.year_of_experience}")

    def update_year_of_experience(self, years):
        self.year_of_experience += years
        print(f"Updated Year of Experience: {self.year_of_experience}")

    def increase_salary(self, amount):
        self.salary += amount
        print(f"Updated Salary: {self.salary}")


employee101 = Employee("Ali", "03001234567", "Karachi", 50000, "Developer", 2020, "BS Computer Science", 4)

print(employee101.name)
print(employee101.number)
print(employee101.address)
print(employee101.salary)

employee101.describe_employee()
employee101.increase_salary(10000)
employee101.update_year_of_experience(2)

employees = [
    Employee("Sara", "03011111111", "Lahore", 60000, "Manager", 2018, "MBA", 6),
    Employee("Sana", "03022222222", "Islamabad", 55000, "Analyst", 2019, "BS Statistics", 5),
    Employee("Faiz", "03033333333", "Karachi", 70000, "Team Lead", 2017, "MS Software Engineering", 7),
    Employee("Aliyan", "03044444444", "Multan", 45000, "Developer", 2021, "BS Computer Science", 3),
    Employee("Khushboo", "03055555555", "Hyderabad", 52000, "Designer", 2020, "BS Design", 4),
    Employee("Talha", "03066666666", "Quetta", 48000, "Tester", 2022, "BS IT", 2),
    Employee("Naveed", "03077777777", "Peshawar", 65000, "Developer", 2016, "MS Computer Science", 8),
    Employee("Raheel", "03088888888", "Sukkur", 40000, "Intern", 2023, "BS Computer Science", 1),
    Employee("Alizy", "03099999999", "Larkana", 75000, "Senior Developer", 2015, "MS Software Engineering", 9),
    Employee("Javaid", "03000000000", "Bahawalpur", 58000, "Developer", 2019, "BS Software Engineering", 5),
]

for emp in employees:
    emp.describe_employee()
    print("---")

employees[0].increase_salary(5000)
employees[0].update_year_of_experience(1)

employees[3].increase_salary(8000)
employees[3].update_year_of_experience(2)

employees[7].increase_salary(3000)
employees[7].update_year_of_experience(1)
```
